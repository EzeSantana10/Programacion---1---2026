## `date`, `datetime` y `timedelta`

El módulo `datetime` trae tres clases principales:

```python
from datetime import date, datetime, timedelta

date(2026, 8, 13)             # solo fecha: 13 de agosto de 2026
datetime(2026, 8, 13, 14, 30) # fecha y hora: 13 de agosto de 2026, 14:30
timedelta(days=10, hours=3)   # una duración: 10 días y 3 horas
```

`date.today()` y `datetime.now()` devuelven la fecha (y hora) actual.

Ambas clases tienen atributos `.year`, `.month`, `.day`, y el método `.weekday()`, que devuelve el día
de la semana como número: **0 = lunes ... 6 = domingo**.

```python
d = date(2026, 8, 13)
d.weekday()   # 3 (jueves)
```

Sumando o restando un `timedelta` a una fecha se obtiene otra fecha:

```python
date(2026, 8, 13) + timedelta(days=5)   # 2026-08-18
```

Y restando dos fechas se obtiene un `timedelta`:

```python
datetime(2026, 9, 7, 18, 9) - datetime(2026, 8, 9, 7, 55)   # 29 days, 10:14:00
```

## Formatear fechas con `strftime`

`strftime` convierte una fecha en un string con el formato que se le indique:

```python
d = date(2026, 8, 13)
d.strftime('%d/%m/%Y')   # '13/08/2026'
```

Los códigos más usados:

| Código | Significado          | Ejemplo |
|--------|-----------------------|---------|
| `%d`   | día (2 dígitos)        | `13`    |
| `%m`   | mes (2 dígitos)        | `08`    |
| `%Y`   | año (4 dígitos)        | `2026`  |
| `%A`   | nombre del día         | `Thursday` |
| `%B`   | nombre del mes         | `August`   |

**Ojo:** `%A` y `%B` dan el nombre en **inglés**, salvo que se configure el `locale` del sistema a
español (y no siempre está disponible en Jupyter). Para armar textos en español lo más simple y
portable es tener tus propias listas:

```python
dias_semana = ['lunes', 'martes', 'miércoles', 'jueves', 'viernes', 'sábado', 'domingo']
meses = ['enero', 'febrero', 'marzo', 'abril', 'mayo', 'junio',
         'julio', 'agosto', 'septiembre', 'octubre', 'noviembre', 'diciembre']

dias_semana[d.weekday()]   # 'jueves'
meses[d.month - 1]         # 'agosto'
```

## 11

````python
Lunes = []
timedel = timedelta(days=1)
veiteveiteseis = date(2026, 1, 1)

while veiteveiteseis != date(2027, 1, 1):
    if veiteveiteseis.weekday() == 0:
        Lunes.append(f"Lunes {veiteveiteseis.day} del {veiteveiteseis.month} de {veiteveiteseis.year}\n")
    veiteveiteseis += timedel
print("".join(Lunes))
````
