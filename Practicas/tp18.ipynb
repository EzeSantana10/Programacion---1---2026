7. Encuentre los índices de los elementos distintos de cero en [1,2,0,0,4,0]

```python
lista = [1, 2, 0, 0, 4, 0]
lista_posiciones = []
posicion = 0

for i in range(len(lista)):
    if lista[posicion] != 0:
        lista_posiciones.append(posicion)
        posicion += 1
    else:
        posicion += 1
print(lista_posiciones)



```

9.  Cree un array 10x10 con valores aleatorios y encuentre los valores mínimo y máximo

```python
import numpy as np

array = np.random.randint(0, 100,(10, 10))
print(array)

minn = array[0, 0]
maxx = array[0, 0]

for fila in array:
    for x in fila:
        if x > maxx:
            maxx = x
        if x < minn:
            minn = x
print(maxx, minn)
```
